# Scikit-Learn Basics

This notebook introduces the basic workflow of Scikit-Learn.

Scikit-Learn is an open-source Python machine learning library that
provides implementations of many machine learning algorithms,
preprocessing techniques, model evaluation tools, and utilities.

The main workflow used in Scikit-Learn is:

1. Import
2. Prepare data
3. Create a model
4. Fit the model
5. Predict
6. Evaluate

This notebook focuses on understanding the Scikit-Learn API and
the meaning of `fit()`, `predict()`, and model parameters.

## 1. Importing Scikit-Learn

Scikit-Learn is usually imported using:

    from sklearn.module import Class

Here:

- `sklearn` → main package
- `module` → specific part of Scikit-Learn
- `Class` → algorithm or utility we want to use

Example:

    from sklearn.linear_model import LinearRegression

`linear_model` is a module and `LinearRegression` is a class.

```python
from sklearn.linear_model import LinearRegression
```

## 2. Creating a Model

After importing a model, we create an object from its class.

Example:

    model = LinearRegression()

At this point, the model has been created but it has NOT been trained.

Training happens later using the `fit()` method.

```python
model = LinearRegression()

print(model)
```

## 3. Preparing Data

For supervised learning, we normally have:

- `X` → input features
- `y` → target/output

Here we create a simple dataset where:

    fw,b(x) = w*x+b --> model function for linear regression
    y = 2x + 1

The model should learn this relationship from the examples.

```python
import numpy as np

X = np.array([
    [1],
    [2],
    [3],
    [4],
    [5]
])

y = np.array([
    3,
    5,
    7,
    9,
    11
])

print(X)
print(y)
```

## 4. Understanding the Shape of X and y

`X` contains the input features.

For one feature, X still needs to be a 2-dimensional array:

    X.shape = (number_of_samples, number_of_features)

Here:

    X.shape = (5, 1)

This means:

- 5 training examples
- 1 feature

The target `y` contains one output for each example.

```python
print("X shape:", X.shape)
print("y shape:", y.shape)
```

## 5. Training the Model with fit()

The `fit()` method trains the model.

    model.fit(X, y)

General meaning:

    fit() = learn from the training data

For LinearRegression, the model learns:

- coefficients (weights)
- intercept (bias)

The model tries to find parameters that best represent the
relationship between X and y.

```python
model.fit(X, y)
```

## 6. Viewing Learned Parameters

After `fit()`, the model has learned parameters.

For LinearRegression:

- `coef_` → learned coefficient/weight
- `intercept_` → learned intercept/bias

The model has learned approximately:

    y = 2x + 1

    here fw,b(x) = y
      w = 2
      b = 1

```python
print("Coefficient:", model.coef_)
print("Intercept:", model.intercept_)
```

## 7. Making Predictions with predict()

After training, we can use `predict()`.

    model.predict(X)

`predict()` uses the parameters learned during `fit()` to
produce predictions for new input data.

```python
y_pred = model.predict(X)

print("Predictions:", y_pred)
```

## 8. Prediction on New Data

The model has already learned the relationship between X and y.

Now we can provide new values that the model has not seen before.

For example:

    x = 6

The model should predict approximately:

    y = 2(6) + 1 = 13

```python
X_new = np.array([
    [6],
    [7],
    [8]
])

y_new_pred = model.predict(X_new)

print("Predictions:", y_new_pred)
```

## 9. fit() vs predict()

### fit()

    model.fit(X, y)

Means:

    Learn from the data.

The model uses X and the known target y to learn its parameters.

### predict()

    model.predict(X_new)

Means:

    Use the learned model to make predictions.

During prediction, we provide the input X but normally do not
provide the target y.

Simple workflow:

    X_train + y_train
            ↓
          fit()
            ↓
      trained model
            ↓
        predict()
            ↓
       predictions

## 10. Evaluating the Model

Scikit-Learn models often provide a `score()` method.

For `LinearRegression`, `score()` returns the R² score.

R² measures how well the model explains the variation in the
target values.

A value closer to 1 generally indicates that the model explains
the training data well.

Note: The exact scoring method depends on the estimator.

```python
score = model.score(X, y)

print("R² Score:", score)  #
```

## 11. Complete Scikit-Learn Workflow

The basic Scikit-Learn workflow is:

    1. Import
    2. Prepare data
    3. Create model
    4. Fit
    5. Predict
    6. Evaluate

```python
from sklearn.linear_model import LinearRegression
import numpy as np
from sklearn.preprocessing import StandardScaler


x = np.array([ [1],[2],[3],[4],[5] ])
y = np.array([5,3,7,9,11])

# create model

model = LinearRegression()

# train model
model.fit(x,y)

# 4. Make predictions
X_new = np.array([[6], [7], [8]])
predictions = model.predict(X_new)

# 5) score

score = model.score(x,y)

print("Coefficient:", model.coef_)
print("Intercept:", model.intercept_)
print("Predictions:", predictions)
print("R² Score:", score)
```

## 13. Important Scikit-Learn Terminology

### Package

`sklearn`

The main Scikit-Learn package.

### Module

Examples:

    sklearn.linear_model
    sklearn.preprocessing
    sklearn.metrics

A module contains related functionality.

### Class

Examples:

    LinearRegression
    StandardScaler
    LogisticRegression

A class provides the implementation.

### Object

Example:

    model = LinearRegression()

`model` is an object/instance of `LinearRegression`.

### Method

Examples:

    model.fit()
    model.predict()
    model.score()

Methods perform operations using the model object.

### Attribute

Examples:

    model.coef_
    model.intercept_

Attributes contain information stored in the model.

## 14. fit(), transform(), and predict()

### fit()

Learn from data.

    object.fit(data)

### transform()

Apply a learned transformation.

    object.transform(data)

### fit_transform()

Learn the transformation and apply it.

    object.fit_transform(data)

### predict()

Use a trained model to produce predictions.

    model.predict(data)

These methods form an important part of the Scikit-Learn API.

## 15. From Scratch vs Scikit-Learn

In the previous Linear Regression implementation, the algorithm
was implemented manually.

From scratch, we implemented concepts such as:

- Cost function
- Gradient calculation
- Gradient Descent
- Weight initialization
- Bias initialization
- Feature normalization

Scikit-Learn provides ready-to-use implementations of many
machine learning algorithms.

Instead of manually implementing every part of the algorithm,
we can use the Scikit-Learn API.

For example:

From scratch:

    manually calculate gradients
    manually update w and b
    repeat for many iterations

Scikit-Learn:

    model = LinearRegression()
    model.fit(X, y)
    predictions = model.predict(X)

The purpose of the from-scratch implementation is to understand
the mathematics and algorithm.

The purpose of Scikit-Learn is to efficiently apply established
machine learning algorithms using a consistent API.
