# 03 - MinMaxScaler

MinMaxScaler is another feature scaling technique in Scikit-Learn.

It changes values to a fixed range. By default, the range is **0 to 1**.

In this notebook we will learn `fit()`, `transform()`, `fit_transform()`, train/test scaling, and `inverse_transform()`.

## 1. Why do we need MinMaxScaler?

Different features can have very different ranges.

For example:
- House size: 500 to 5000
- Bedrooms: 1 to 5
- Age: 0 to 50

MinMaxScaler brings each feature to a similar range. This can help models that are sensitive to feature scale, especially gradient-descent-based models.

## 2. Min-Max formula

$$
x_{scaled} = \frac{x-x_{min}}{x_{max}-x_{min}}
$$

Where:
- `x` = original value
- `x_min` = minimum value
- `x_max` = maximum value

Example: if values are from 10 to 50 and `x = 30`:

$$
x_{scaled}=\frac{30-10}{50-10}=0.5
$$

````python
import numpy as np
from sklearn.preprocessing import MinMaxScaler
````

## 3. Simple example

Let's start with one feature: house size.

````python
X = np.array([
    [1000],
    [1500],
    [2000],
    [2500],
    [3000]
])

print(X)
````

````python
scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)
print(np.round(X_scaled, 2))
````

### What happened?

The minimum is `1000` and the maximum is `3000`.

| Original | Scaled |
|---:|---:|
| 1000 | 0.00 |
| 1500 | 0.25 |
| 2000 | 0.50 |
| 2500 | 0.75 |
| 3000 | 1.00 |

The smallest training value becomes `0` and the largest becomes `1`.

## 4. Understanding `fit()`

`fit()` learns the minimum and maximum values from the data. It does not transform the data yet.

````python
scaler = MinMaxScaler()
scaler.fit(X)

print("Minimum:", scaler.data_min_)
print("Maximum:", scaler.data_max_)
````

## 5. Understanding `transform()`

`transform()` uses the minimum and maximum learned by `fit()` and converts the original values to scaled values.

So:
- `fit()` = learn min and max
- `transform()` = apply the scaling

````python
scaler = MinMaxScaler()
scaler.fit(X)
X_scaled = scaler.transform(X)
print(np.round(X_scaled, 2))
````

## 6. MinMaxScaler with multiple features

MinMaxScaler scales each column separately.

````python
X = np.array([
    [1000, 2, 10],
    [1500, 3, 20],
    [2000, 4, 30],
    [2500, 5, 40],
    [3000, 6, 50]
])

scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)
print(np.round(X_scaled, 2))
````

The three columns are scaled separately. Each feature gets its own minimum and maximum.

## 7. Train-test data: very important

Just like with `StandardScaler`, split the data first.

```python
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

We do **not** fit the scaler on `X_test`. The scaler learns the minimum and maximum only from training data, then uses those same values for test data. This prevents data leakage.

````python
from sklearn.model_selection import train_test_split

X = np.array([
    [1000], [1200], [1400], [1600], [1800],
    [2000], [2200], [2400], [2600], [2800]
])

X_train, X_test = train_test_split(X, test_size=0.2, random_state=42)

scaler = MinMaxScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Training data:")
print(np.round(X_train_scaled, 2))
print("\nTesting data:")
print(np.round(X_test_scaled, 2))
````

### Important

A test value can sometimes be less than `0` or greater than `1` if it is outside the training minimum and maximum. That is normal because the scaler uses the training range.

## 8. MinMaxScaler vs StandardScaler

| StandardScaler | MinMaxScaler |
|---|---|
| Uses mean and standard deviation | Uses minimum and maximum |
| `(x - mean) / std` | `(x - min) / (max - min)` |
| Usually mean ≈ 0 and std ≈ 1 | Training values normally lie between 0 and 1 |
| Can produce negative values | Normally produces 0 to 1 for training data |

Neither scaler is always the best choice. The choice depends on the data and the model.
